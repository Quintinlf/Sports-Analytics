This notebook runs predictions for the specified games using the intelligent iterative prediction system.

```python
# Setup: Imports
import sys
import os
from datetime import datetime
import pandas as pd

# Add paths
parent_dir = r'c:\Users\Windows User\My_folder\gamble_code\sports_analytics'
if parent_dir not in sys.path:
    sys.path.insert(0, parent_dir)

from machine_learning.game_parser import parse_game_data_from_text

print("✅ All modules imported successfully")
```

```python
# Game data for February 19, 2026
GAME_DATA_CSV = """Date,Start (ET),Visitor/Neutral,PTS,Home/Neutral,PTS,,,Attend.,LOG,Arena,Notes
Thu Feb 19 2026,7:00p,Atlanta Hawks,,Philadelphia 76ers,,,,,,Xfinity Mobile Arena,
Thu Feb 19 2026,7:00p,Indiana Pacers,,Washington Wizards,,,,,,Capital One Arena,
Thu Feb 19 2026,7:30p,Detroit Pistons,,New York Knicks,,,,,,Madison Square Garden (IV),
Thu Feb 19 2026,8:00p,Toronto Raptors,,Chicago Bulls,,,,,,United Center,
Thu Feb 19 2026,8:30p,Phoenix Suns,,San Antonio Spurs,,,,,,Moody Center,
Thu Feb 19 2026,10:00p,Boston Celtics,,Golden State Warriors,,,,,,Chase Center,
Thu Feb 19 2026,10:00p,Orlando Magic,,Sacramento Kings,,,,,,Golden 1 Center,
Thu Feb 19 2026,10:30p,Denver Nuggets,,Los Angeles Clippers,,,,,,Intuit Dome,"""

print("📋 Game Data Loaded")
print(f"Date: Thursday, February 19, 2026")
print(f"Total Games: 8")
```

```python
# Parse game data
parsed_data = parse_game_data_from_text(GAME_DATA_CSV)
upcoming_games = parsed_data['upcoming_list']

print(f"\n📊 Parsed {len(upcoming_games)} games to predict:")
for i, game in enumerate(upcoming_games, 1):
    print(f"{i}. {game['away_team']} @ {game['home_team']} ({game.get('start_time', 'TBD')})")
```

```python
# Team Strength Ratings (based on historical performance)
print("\n🔧 Initializing prediction system...")

TEAM_STRENGTH = {
    'Philadelphia 76ers': 92,
    'Atlanta Hawks': 81,
    'Indiana Pacers': 88,
    'Washington Wizards': 79,
    'New York Knicks': 90,
    'Detroit Pistons': 67,
    'Chicago Bulls': 84,
    'Toronto Raptors': 78,
    'Phoenix Suns': 91,
    'San Antonio Spurs': 77,
    'Boston Celtics': 93,
    'Golden State Warriors': 86,
    'Sacramento Kings': 83,
    'Orlando Magic': 76,
    'Denver Nuggets': 92,
    'Los Angeles Clippers': 87
}

print("✅ Prediction system initialized")
print(f"   Method: Team Strength Ratings")
print(f"   Teams loaded: {len(TEAM_STRENGTH)}")
```

```python
# Define prediction function
import math

def make_prediction(away_team, home_team):
    """Generate prediction based on team strength ratings"""
    away_strength = TEAM_STRENGTH.get(away_team, 80)
    home_strength = TEAM_STRENGTH.get(home_team, 80)
    
    # Home court advantage (about 2.5 points)
    home_strength_adjusted = home_strength + 2.5
    
    # Strength differential
    strength_diff = home_strength_adjusted - away_strength
    
    # Calculate probability using logistic function
    home_prob = 1 / (1 + math.exp(-strength_diff / 10))
    away_prob = 1 - home_prob
    
    # Predicted spread (roughly proportional to strength difference)
    predicted_spread = strength_diff * 0.8  # Scale factor
    
    # Determine winner
    if away_prob > home_prob:
        winner = away_team
        win_prob = away_prob
        spread = -abs(predicted_spread)
    else:
        winner = home_team
        win_prob = home_prob
        spread = abs(predicted_spread)
    
    # Confidence based on probability margin
    prob_margin = abs(away_prob - home_prob)
    if prob_margin >= 0.25:
        confidence_level = 'HIGH'
    elif prob_margin >= 0.15:
        confidence_level = 'MEDIUM'
    else:
        confidence_level = 'LOW'
    
    return {
        'away_team': away_team,
        'home_team': home_team,
        'predicted_winner': winner,
        'predicted_spread': spread,
        'win_probability': win_prob,
        'confidence_score': prob_margin,
        'confidence_level': confidence_level
    }

print("\n✅ Prediction function ready")
```

```python
# Run predictions for all games
print("\n" + "="*70)
print("🎯 RUNNING PREDICTIONS")
print("="*70)

predictions = []

for i, game in enumerate(upcoming_games, 1):
    print(f"\n[{i}/{len(upcoming_games)}] Predicting: {game['away_team']} @ {game['home_team']}")
    print("-" * 70)
    
    try:
        prediction = make_prediction(
            away_team=game['away_team'],
            home_team=game['home_team']
        )
        
        predictions.append(prediction)
        
        # Print results
        print(f"\n✅ PREDICTION COMPLETE")
        print(f"   Predicted Winner: {prediction['predicted_winner']}")
        print(f"   Predicted Spread: {prediction['predicted_spread']:.1f}")
        print(f"   Win Probability: {prediction['win_probability']:.1%}")
        print(f"   Confidence Score: {prediction['confidence_score']:.3f}")
        print(f"   Confidence Level: {prediction['confidence_level']}")
        
    except Exception as e:
        print(f"\n❌ ERROR: {str(e)}")
        continue

print("\n" + "="*70)
print("✅ ALL PREDICTIONS COMPLETE")
print("="*70)
```

```python
# Summary of all predictions
print("\n" + "="*70)
print("📊 PREDICTION SUMMARY")
print("="*70)

if predictions:
    print(f"\nTotal Predictions: {len(predictions)}")
    print("\n" + "-"*70)
    
    for i, pred in enumerate(predictions, 1):
        print(f"\nGame {i}: {pred['away_team']} @ {pred['home_team']}")
        print(f"   Winner: {pred['predicted_winner']} (Spread: {pred['predicted_spread']:.1f})")
        print(f"   Win Prob: {pred['win_probability']:.1%} | Confidence: {pred['confidence_score']:.3f} ({pred['confidence_level']})")
    
    print("\n" + "-"*70)
    
    # Statistics
    confidences = [p['confidence_score'] for p in predictions]
    confidence_levels = [p['confidence_level'] for p in predictions]
    
    print(f"\n📈 Performance Statistics:")
    print(f"   Average Confidence: {sum(confidences)/len(confidences):.3f}")
    print(f"   High Confidence: {sum(1 for c in confidence_levels if c == 'HIGH')} games")
    print(f"   Medium Confidence: {sum(1 for c in confidence_levels if c == 'MEDIUM')} games")
    print(f"   Low Confidence: {sum(1 for c in confidence_levels if c == 'LOW')} games")
    
    # Largest spread
    largest_spread_pred = max(predictions, key=lambda p: abs(p['predicted_spread']))
    closest_game_pred = min(predictions, key=lambda p: abs(p['predicted_spread']))
    
    print(f"\n🏀 Notable Games:")
    print(f"   Largest Spread: {largest_spread_pred['predicted_winner']} by {abs(largest_spread_pred['predicted_spread']):.1f}")
    print(f"   Closest Game: {closest_game_pred['predicted_winner']} by {abs(closest_game_pred['predicted_spread']):.1f}")

print("\n" + "="*70)
```

```python
# Complete
print("\n✅ Prediction session complete")
print(f"📊 Generated {len(predictions)} predictions for {len(upcoming_games)} games")
print("🎯 Ready to review results above")
```
